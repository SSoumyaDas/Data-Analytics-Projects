# Matches by season

```python
matches_by_season = df['season'].value_counts().sort_index()

matches_by_season
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(
    matches_by_season.index,
    matches_by_season.values,
    marker='o'
)

plt.title('IPL Matches by Season')
plt.xlabel('Season')
plt.ylabel('Number of Matches')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Team wins

```python
team_wins = df['winner'].value_counts()

team_wins
```

```python
plt.figure(figsize=(10, 6))

team_wins.sort_values().plot(kind='barh')

plt.title('IPL Wins by Team')
plt.xlabel('Number of Wins')
plt.ylabel('Team')
plt.tight_layout()
plt.show()
```

```python
matches_played = (
    pd.concat([df['team1'], df['team2']])
    .value_counts()
)

team_performance = pd.DataFrame({
    'matches_played': matches_played,
    'wins': team_wins
})

team_performance['win_percentage'] = (
    team_performance['wins'] / team_performance['matches_played'] * 100
)

team_performance.sort_values(
    'win_percentage',
    ascending=False
)
```

```python
win_pct = team_performance.sort_values('win_percentage')

plt.figure(figsize=(10, 6))

plt.barh(
    win_pct.index,
    win_pct['win_percentage']
)

plt.title('Team Win Percentage')
plt.xlabel('Win Percentage (%)')
plt.ylabel('Team')
plt.tight_layout()
plt.show()
```

# Does winning the toss help?

```python
toss_impact = df['toss_winner_won'].value_counts()

toss_impact
```

```python
toss_win_percentage = (
    df['toss_winner_won']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

toss_win_percentage
```

```python
plt.figure(figsize=(7, 5))

toss_win_percentage.plot(kind='bar')

plt.title('Toss Winner vs Match Winner')
plt.xlabel('Toss Winner Won Match')
plt.ylabel('Percentage of Matches')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
toss_decision = df['toss_decision'].value_counts()

toss_decision
```

```python
toss_decision_percentage = (
    df['toss_decision']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

toss_decision_percentage
```

```python
plt.figure(figsize=(7, 5))

toss_decision_percentage.plot(kind='bar')

plt.title('Toss Decision')
plt.xlabel('Decision')
plt.ylabel('Percentage of Matches')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Toss decision vs match outcome

```python
toss_decision_outcome = pd.crosstab(
    df['toss_decision'],
    df['toss_winner_won']
)

toss_decision_outcome
```

```python
toss_decision_outcome_pct = pd.crosstab(
    df['toss_decision'],
    df['toss_winner_won'],
    normalize='index'
).mul(100).round(2)

toss_decision_outcome_pct
```
